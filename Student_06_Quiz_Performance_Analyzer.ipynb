## 4. NumPy practice

```python
# TODO: Select numeric columns relevant to your project
# TODO: Convert them to a NumPy array using .to_numpy()
# TODO: Calculate average, maximum and minimum

quiz_cols = ['Quiz_1', 'Quiz_2', 'Quiz_3', 'Quiz_4']
quiz_array = df[quiz_cols].to_numpy()

print(f"Overall Average Score: {np.mean(quiz_array):.2f}")
print(f"Highest Single Score:  {np.max(quiz_array)}")
print(f"Lowest Single Score:   {np.min(quiz_array)}")
```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
# TODO: Create calculated column(s)
# TODO: Filter records using a condition
# TODO: Sort records and display the top/bottom results

df['Total_Score'] = df[quiz_cols].sum(axis=1)
df['Average_Score'] = df[quiz_cols].mean(axis=1).round(2)

high_scorers = df[df['Average_Score'] >= 80]
print(f"Students with Average >= 80 ({len(high_scorers)} students):")
display(high_scorers[['Student_ID', 'Name', 'Average_Score']])

df_sorted = df.sort_values(by='Average_Score', ascending=False)
print("Top 3 Performers:")
display(df_sorted[['Student_ID', 'Name', 'Total_Score', 'Average_Score']].head(3))
print("Lowest Performer:")
display(df_sorted[['Student_ID', 'Name', 'Total_Score', 'Average_Score']].tail(1))
```
